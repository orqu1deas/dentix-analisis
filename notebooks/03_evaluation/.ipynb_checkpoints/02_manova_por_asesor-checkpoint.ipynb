##Dataset y liberías

```python
import numpy as np
import pandas as pd
from statsmodels.multivariate.manova import MANOVA
from sklearn.preprocessing import StandardScaler
```

```python
#Dataset
df=pd.read_csv("df_clean.csv")
df.head(10)
```

##Escalamiento de datos

```python
cols = ["SCORE", "DIAS DE MORA ", "SALDO VENCIDO", "CUOTAMENSUAL","TOTAL INGRESOS","TOTAL EGRESOS"]
df[cols] = StandardScaler().fit_transform(df[cols])
```

##Modelo

```python
#Definir variables dependientes
Y = df[["SCORE","DIAS DE MORA ","SALDO VENCIDO","CUOTAMENSUAL","TOTAL INGRESOS","TOTAL EGRESOS"]]
```

```python
#Variable independiente
df["CLINICA"] = df["CLINICA"].astype("category")
X = df[["CLINICA"]]
```

```python
manova = MANOVA.from_formula(
    "SCORE + Q('DIAS DE MORA ') + Q('SALDO VENCIDO') + CUOTAMENSUAL + Q('TOTAL INGRESOS') + Q('TOTAL EGRESOS') ~ C(CLINICA)",
    data=df
)

print(manova.mv_test())
```
